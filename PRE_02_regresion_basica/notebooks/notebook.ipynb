```python
import pandas as pd

dataset = pd.read_csv("../data/auto_mpg.csv")
dataset.head()
```

```python
dataset.isna().sum()
```

```python
dataset = dataset.dropna()
dataset.isna().sum()
```

```python
dataset.Origin.value_counts()
```

```python
dataset["Origin"] = dataset["Origin"].map(
    {   
        1: "USA", 
        2: "Europe", 
        3: "Japan",
    },
)
dataset.head(20)
```

```python
dataset = pd.get_dummies(dataset, columns=["Origin"], prefix="", prefix_sep="")
dataset.head()
```
